```python
from sqlalchemy import create_engine, text
import bcrypt
import dotenv
import os
```

```python
dotenv.load_dotenv("db_palete_caixa.env")
```

```python
HOST = os.getenv("HOST")
PORT = os.getenv("PORT")
USER = os.getenv("USER")
DATABASE = os.getenv("DATABASE")
PASS = os.getenv("PASS")
print(HOST, PORT, USER, DATABASE)
```

```python
if PASS:
    print(f"{'\u2022' * 5}")
else:
    print("Senha nao encontrada")
```

```python
url = f"postgresql+psycopg2://{USER}:{PASS}@{HOST}:{PORT}/{DATABASE}"
```

```python
engine = create_engine(
    url, connect_args={"sslmode": "prefer"}
)
```

```python
import pandas as pd
df_tabelas = pd.read_sql("""
        SELECT table_name
        FROM information_schema.tables
        WHERE table_schema = 'public'
          AND table_name IN ('clientes', 'tipos_status', 'tipos_usuarios')
        ORDER BY table_name;
    """, engine)
df_tabelas
```
